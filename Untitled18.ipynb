```python
import pandas as pd

# Updated the path to use the correct filename found in the extracted directory
df = pd.read_csv("drug_discovery_data/ai_drug_discovery_molecule_property_prediction.csv")

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python
# 1. Check the dataset size
print("Dataset shape:", df.shape)

# 2. Inspect data types and missing values
df.info()

# 3. Count missing values in each column
print("\nMissing values:")
print(df.isnull().sum())

# 4. Check for duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# 5. Get summary statistics for numerical columns
display(df.describe().round(2))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Select important molecular properties
properties = [
    "molecular_weight",
    "logP",
    "num_hydrogen_bond_donors",
    "num_hydrogen_bond_acceptors",
    "topological_polar_surface_area",
    "rotatable_bonds",
    "aromatic_rings"
]

# Calculate average values
print("Average molecular properties:")
display(df[properties].mean().round(2).to_frame("Mean"))
# Plot molecular weight distribution
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="molecular_weight", bins=30, kde=True)
plt.title("Distribution of Molecular Weight")
plt.xlabel("Molecular Weight (g/mol)")
plt.ylabel("Number of Molecules")
plt.show()
```

```python
# Summarize predicted binding affinity
print(df["binding_affinity_kcal_mol"].describe().round(2))

# Compare binding affinity with molecular weight
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="molecular_weight",
    y="binding_affinity_kcal_mol",
    alpha=0.6
)
plt.title("Molecular Weight vs Predicted Binding Affinity")
plt.xlabel("Molecular Weight (g/mol)")
plt.ylabel("Predicted Binding Affinity (kcal/mol)")
plt.show()
```

```python
# Count molecules in each toxicity category
toxicity_counts = df["toxicity_risk_level"].value_counts()

print(toxicity_counts)

# Visualize the categories
toxicity_counts.plot(
    kind="bar",
    figsize=(7, 4)
)

plt.title("Distribution of Predicted Toxicity Risk")
plt.xlabel("Toxicity Risk Level")
plt.ylabel("Number of Molecules")
plt.xticks(rotation=0)
plt.show()
```

```python
# Compare molecular properties by drug-likeness label
display(
    df.groupby("drug_likeness")[properties]
      .mean()
      .round(2)
)

# Count Yes and No labels
print("\nDrug-likeness counts:")
print(df["drug_likeness"].value_counts())
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

numeric_df = df.select_dtypes(include="number")

correlation = numeric_df.corr()

plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    center=0
)

plt.title("Correlation Between Molecular Features")
plt.tight_layout()
plt.show()
```

```python
# Sort molecules by predicted binding affinity
top_molecules = df.sort_values(
    by="binding_affinity_kcal_mol",
    ascending=True
)

# Display the 10 molecules with the most negative values
display(
    top_molecules[
        ["molecule_id", "molecular_weight", "logP",
         "bioactivity_score", "binding_affinity_kcal_mol",
         "toxicity_risk_level", "drug_likeness"]
    ].head(10)
)
```

```python
toxicity_analysis = df.groupby("toxicity_risk_level").agg(
    molecule_count=("molecule_id", "count"),
    average_molecular_weight=("molecular_weight", "mean"),
    average_logP=("logP", "mean"),
    average_bioactivity=("bioactivity_score", "mean"),
    average_binding_affinity=("binding_affinity_kcal_mol", "mean")
)

display(toxicity_analysis.round(3))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="logP",
    y="bioactivity_score",
    hue="toxicity_risk_level",
    alpha=0.7
)

plt.title("Lipophilicity vs Bioactivity Score")
plt.xlabel("logP")
plt.ylabel("Bioactivity Score")
plt.tight_layout()
plt.show()
```

```python
numeric_df = df.select_dtypes(include="number").drop(
    columns=["molecule_id"], errors="ignore"
)

correlation = numeric_df.corr()

# Extract each unique pair of variables
pairs = correlation.where(
    __import__("numpy").triu(
        __import__("numpy").ones(correlation.shape), k=1
    ).astype(bool)
).stack().reset_index()

pairs.columns = ["Property 1", "Property 2", "Correlation"]

print("Strongest positive relationships:")
display(pairs.sort_values("Correlation", ascending=False).head(5))

print("Strongest negative relationships:")
display(pairs.sort_values("Correlation", ascending=True).head(5))
```
